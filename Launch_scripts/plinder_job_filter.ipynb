```python
import pandas as pd

df_incomplete = pd.read_csv('plinder_launch_incomplete.csv')

# 165 randomly selected protein-ligand systems
protein_ligand = df_incomplete.sample(n=165, random_state=42).reset_index(drop=True)

# 100 randomly selected protein-only systems (ligand code blank)
protein_only = df_incomplete.sample(n=100, random_state=7).reset_index(drop=True)
protein_only['ligand_unique_ccd_code'] = ''

combined = pd.concat([protein_ligand, protein_only], ignore_index=True)

print(f'Total systems: {len(combined)}')
print(f'Protein-ligand: {len(protein_ligand)}')
print(f'Protein-only:   {len(protein_only)}')
combined.head(10)
```

```python
combined.to_csv('plinder_job_filter_output.csv', index=False)
print('Saved to plinder_job_filter_output.csv')
```
